## Data Cleaning Process for Dataset

#### Analyze the data before cleaning

```python
# import the dataset 
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
# see # of rows, columns, Dtypes, null values
df.info()
```

```python
# Look at the summary statistics of the dataset (numerical values unless you put include='all')
df.describe(include='all')
```

```python
# check missing values

'''
We see col "Review Rating" has 37 NA values so:
1. Understanding why the data is missing
2. Evaluate how much is missing 
    If it's less than 5% & random = replace with median (for outliers)
    If it's between 5-30 = replace with median (for numeric) (mode/"unknown" for categorical)
    If it's greater than 30% = assess importance--consider dropping col 
'''

df.isnull().sum()
```

#### Clean the Data 

```python
# fills the missing values in col "Review Rating" with the median by their category group
# .transform() is making sure the rows are not collapsed like Window function in SQL
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Check to see if the null have been replaced
df.isnull().sum()
```

```python
# Column name manipulation 

# turn to lower case 
df.columns = df.columns.str.lower()

# add the _
df.columns = df.columns.str.replace(' ', '_')

# change purchase_amount_(usd) with purchase_amount
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
df.head()
```

```python
# create a col age_group 
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

'''
 .qcut = automatically divides the data into equal-sized percentile buckets. 
q=4 means: from a range of 18-70 age-group: 
1. the first 25% will be in bucket 1
2. the second 25% will be in bucket 2 
3. the third 25% will be in bucket 3
4. the fourth 25% will be in bucket 4

Additionally:
.cut - allows you to choose the range so:
1. put the 18-35 in bucket 1
2. put the 35-45 in bucket 2
'''
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
'''
create col called purchase frequence days so we can create numeric
stats for the 'frequency_of_purchases' col instead of worrying about
the text it gave us in that col 
'''
# first create a dict for each frequency_of_purchases
frequency_mapping = {
    'Fortnightly': 14, 
    'Weekly': 7, 
    'Annually': 365,
    'Quarterly': 90, 
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Every 3 Months': 90
}

# Then create a new col with the new data of the value that maps to each text 
df['frequency_of_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_days', 'frequency_of_purchases']].head(10)
```

```python
'''
Do we need both columns? If promo is used we can conclude a discount is applied
BUT it's possible automatic discount could be applied ex: during seasons
SO lets see if each of the rows match one another 
'''
df[['discount_applied', 'promo_code_used']]
```

```python
# .all() checkes if each row in 'discount_applied' matches 'promo_code_used'
# if so, return TRUE if not returns FALSE
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# We don't need 'promo_code_used' if they match 
df.drop(columns='promo_code_used', inplace=True)
```
